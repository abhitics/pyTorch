# Forward Hooks

```python
import torch
import torch.nn as nn
def my_pre_hook(module , inputs): #forward pre hook
    print("--Pre-Hook Fired--")
    print(f"The input coming from layer is {inputs}")

def my_hook(module , inputs , outputs): #forward hook
    print("--Hook Fired--")
    print(f"FInal Output generated is {outputs.detach()}") # detach so that it stopped to be tracked

layer=nn.Sequential(
    nn.Linear(5,1),
    nn.ReLU()
)

layer.register_forward_pre_hook(my_pre_hook) # Syntax : register_forward_pre_hook(hook)
layer.register_forward_hook(my_hook) # Syntax : register_forward_hook(hook)

x=torch.tensor([[1.0,2.0,3.0,4.0,7.0]])
out=layer(x)
```
